```python
def format_2d_array(arr):
    max_width = max(len(str(x)) for row in arr for x in row)
    lines = []
    for row in arr:
        formatted = " ".join(f"{str(x).rjust(max_width)}" for x in row)
        lines.append(formatted)
    return "\n".join(lines)
M = [[1,200,3],[40,5,600]]
print(format_2d_array(M))
```

```python
def stats(x):
    a = np.ravel(x)
    mean = np.sum(a)/len(a)
    s = np.sort(a)
    n = len(a)
    median = s[n//2] if n%2==1 else (s[n//2-1]+s[n//2])/2
    v,t = np.unique(a,return_counts=True)
    mode = v[np.argmax(t)]
    return mean,median,mode
A7 = np.random.randint(0,20,(4,4))
B7 = np.random.randint(0,20,(2,8))
C7 = np.random.randint(0,20,(1,12))
print("A7:\n", A7)
print("statsA:", stats(A7))
print("B7:\n", B7)
print("statsB:", stats(B7))
print("C7:\n", C7)
print("statsC:", stats(C7))
```
