```python
s = input("Enter string: ").lower()
vowels="aeiou"
count={v:0 for v in vowels}
for ch in s:
    if ch in vowels:
        count[ch]+=1
print(count)
print("Vowels present:",[v for v in vowels if count[v]>0])
```

```python
mapping = {"0":"Zero","1":"One","2":"Two","3":"Three","4":"Four","5":"Five","6":"Six","7":"Seven","8":"Eight","9":"Nine"}
def num_to_words(n):
    return " ".join(mapping[d] for d in str(n))
print(num_to_words(452))
```

```python
s = input("Enter sentence: ").lower()
words=s.split()
freq={}
for w in words:
    freq[w]=freq.get(w,0)+1
duplicates={w:f for w,f in freq.items() if f>1}
print(duplicates)
```
