```python
cost_meal=float(input("Enter the cost of the meal: "))
tip=0.18*cost_meal
tax=0.05*cost_meal
grand_total=cost_meal+tax+tip
print(f"The cost was: {cost_meal:.2f}")
print(f"The tax was: {tax:.2f}")
print(f"The tip was: {tip:.2f}")
print(f"The Grand Total was: {grand_total:.2f}")
```

```python
num=int(input("Enter a 4 digit integer: "))
dig1=num//1000
dig2=(num//100)%10
dig3=(num//10)%10
dig4=num%10
sum1=dig1+dig2+dig3+dig4
print("The sum is: ",sum1)
```
