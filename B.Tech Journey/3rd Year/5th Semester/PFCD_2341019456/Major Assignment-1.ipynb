1. Write a Python function basic salary that accepts two parameters: hourly rate and hours worked per week.The function should calculate the basic salary per month (assuming a month has 4 weeks). If the hours worked per week exceed 40, create another function overtime salary, where every extra hour worked is paid at 1.5 times the normal hourly rate. Finally, create another function total salary that returns the sum of the basic salary and overtime.

```python
def BasicSalary(HrRate,HrPerWeek):
    return HrRate*HrPerWeek*4

def Overtime_sal(HrRate,HrPerWeek):
    if HrPerWeek>40:
        extra=HrPerWeek-40
        return extra*HrRate*1.5*4
    else:
        return 0

def total_salary(HrRate,HrPerWeek):
    basic=BasicSalary(HrRate,HrPerWeek)
    over=Overtime_sal(HrRate,HrPerWeek)
    return basic+over

print("Enter Employee Details: ")
HrRate=float(input("Hourly Rate (Rs): "))
HrPerWeek=float(input("Hours Worked/Week: "))

basic_salary=BasicSalary(HrRate,HrPerWeek)
overtime_sum=Overtime_sal(HrRate,HrPerWeek)
total_sal=total_salary(HrRate,HrPerWeek)

print("The Rate And Hours Entered Are: ",HrRate," ",HrPerWeek)
print(f"Basic Salary Is: Rs. {basic_salary:.2f}")
print(f"Overtime Salary Is: Rs. {overtime_sum:.2f}")
print(f"Total Salary Will Be: Rs. {total_sal:.2f}")
```

2. Create a function tax amount that shows how much taxes are deducted from the basic salary. Taxes are applied as follows:
- If the salary is less than Rs. 60,000/-, deduct 10% as tax.
- If the salary is between Rs. 60,000/- and Rs. 85,000/-, deduct 15% as tax.
- If the salary is more than Rs. 85,000/-, deduct 20% as tax.

```python
def tax_calculate(basic_sal):
    if basic_sal<60000:
        tax=basic_sal*0.10
    elif 60000<=basic_sal<=85000:
        tax=basic_sal*0.15
    else:
        tax=basic_sal*0.20
    return tax

sal=int(input("Enter Your Basic Salary (Rs): "))
print("Your Entered Basic Salary Is: ",sal)
print("Tax Deducted Is:",tax_calculate(sal))
```

3. Using the function basic salary from Question 1, write another function gross salary that calculates the gross salary of an employee. This function should accept basic salary as input (output from Question 1), consider a fixed value of allowances (e.g., 20% of basic salary), and return the gross salary (basic salary + allowances - tax).

```python
def BasicSalary(HourRate,HourWorkPrWeek):
    return HourRate*HourWorkPrWeek*4

def Overtime_sal(HourRate,HourWorkPrWeek):
    if HourWorkPrWeek>40:
        extra=HourWorkPrWeek-40
        return extra*HourRate*1.5*4
    else:
        return 0

def total_salary(HourRate,HourWorkPrWeek):
    return BasicSalary(HourRate,HourWorkPrWeek)+Overtime_sal(HourRate,HourWorkPrWeek)

def tax_calculate(salary):
    if salary<60000:
        return salary*0.10
    elif salary<=80000:
        return salary*0.15
    else:
        return salary*0.20

def gross_sal(HourRate, HourWorkPrWeek):
    basic=BasicSalary(HourRate,HourWorkPrWeek)
    total=total_salary(HourRate,HourWorkPrWeek)
    allowances= basic*0.20
    tax=tax_calculate(total)
    return basic+allowances-tax

rate=float(input("Enter Hourly Rate (Rs): "))
hours=float(input("Enter Hours Worked/Week: "))
print("The Rate And Hours Entered Are: ",rate," ",hours)
print(f"Gross Salary: Rs. {gross_sal(rate, hours):.2f}")
```

4. Using the gross salary function from Question 3, write a function salary bracket that categorizes the employee’s gross salary into one of the following brackets:
- ”Low income” if gross salary is below Rs. 50,000/-.
- ”Middle income” if gross salary is between Rs. 50,000/- and Rs. 80,000/-.
- ”High income” if gross salary is more than Rs. 80,000/-.

```python
def getSalaryBracket(grossSalaryInput):
    if grossSalaryInput<50000:
        return "Low Income"
    elif grossSalaryInput<=80000:
        return "Middle Income"
    else:
        return "High Income"

userSalary=float(input("Enter Your Gross Salary:"))
print("Your Entered Gross Salary Is: ",userSalary)
incomeLevel=getSalaryBracket(userSalary)
print("Your Salary Bracket Is: ",incomeLevel)
```

5. Take three different sets of employee names, hourly rates and hours worked per week as user input. Write a Python function employee report that generates a formatted report of all employees’ salary details. This function should print the employee names, basic salaries, gross salaries, tax amounts, and salary brackets in a readable format.

```python
def basicSalary(hourlyRate,hoursWorkedPerWeek):
    regularHours=hoursWorkedPerWeek
    if hoursWorkedPerWeek>40:
        regularHours=40
    monthlyBasic=hourlyRate*regularHours*4
    return monthlyBasic

def taxAmount(basicSalaryInput):
    tax=0
    if basicSalaryInput<60000:
        tax = basicSalaryInput*0.10
    elif basicSalaryInput<=85000:
        tax=basicSalaryInput*0.15
    else:
        tax=basicSalaryInput*0.20
    return tax

def grossSalary(basicSalaryInput):
    allowances=basicSalaryInput*0.20
    tax=taxAmount(basicSalaryInput)
    gross=basicSalaryInput+allowances-tax
    return gross

def salaryBracket(grossSalaryInput):
    if grossSalaryInput<50000:
        return "Low Income"
    elif grossSalaryInput<=80000:
        return "Middle Income"
    else:
        return "High Income"
    
def employeeReport():
    employeeData=[]
    print("Three Employee Details:")
    for i in range(3):
        print(f"\nEmployee {i + 1}:")
        name=input("Enter Name: ")
        print("Entered Name: ",name)
        rate=float(input("Enter Hourly Rate: "))
        print("Entered Rate: ",rate)
        hours=float(input("Enter hours worked/week: "))
        print("Entered Hours: ",hours)
        print()
        employeeData.append([name,rate,hours])
    print("\n"+"-"*80)
    print("Employee Salary Report".center(80))
    print("-"*80)
    header=f"{'Name':<20} {'Basic Salary':<15} {'Tax Amount':<15} {'Gross Salary':<15} {'Bracket':<15}"
    print(header)
    print("-"*80)
    for emp in employeeData:
        name=emp[0]
        rate=emp[1]
        hours=emp[2]
        basic=basicSalary(rate, hours)
        tax=taxAmount(basic)
        gross=grossSalary(basic)
        bracket=salaryBracket(gross)
        reportLine=f"{name:<20} {basic:<15.2f} {tax:<15.2f} {gross:<15.2f} {bracket:<15}"
        print(reportLine)
    print("-"*80)
employeeReport()
```
