```python
# Question-1
import matplotlib.pyplot as plt
x=['A','B','C','D']
y1=[10,20,15,25]
y2=[12,18,22,19]
fig,ax=plt.subplots(1,2,figsize=(15,4))
ax[0].bar(x,y1,color='b')
ax[0].set_title("Subplot 1")
ax[1].bar(x,y2,color='c')
ax[1].set_title("Subplot 2")
plt.show()
```

```python
# Question-2
import matplotlib.pyplot as plt
import numpy as np
x=np.arange(4)
y1=[10,20,15,25]
y2=[12,18,22,19]
width=0.4
plt.bar(x-width/2,y1,width,label="Bar 1")
plt.bar(x+width/2,y2,width,label="Bar 2")
plt.xticks(x,['A','B','C','D'])
plt.legend()
plt.show()
```

```python
# Question-3
import matplotlib.pyplot as plt
x=['A','B','C','D']
y=[10,20,15,25]
plt.bar(x,y,color="purple")
plt.title("Standard Bar Graph")
plt.show()
```

```python
# Question-4
import matplotlib.pyplot as plt
import numpy as np
data=np.random.randn(500)
plt.hist(data,bins=20,edgecolor='black')
plt.title("Histogram")
plt.show()
```

```python
# Question-5
import matplotlib.pyplot as plt
labels=['A','B','C','D']
sizes=[15,30,45,10]
colors=['red','green','blue','yellow']
explode=(0,0.1,0,0)
plt.pie(sizes,labels=labels,colors=colors,explode=explode,autopct='%1.1f%%',shadow=True,startangle=90)
plt.title("Pie Chart Example")
plt.show()
```

```python
# Question-9
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,50)
y=np.linspace(-5,5,50)
X,Y=np.meshgrid(x,y)
Z=(X**2+Y**2)**0.5
plt.contour(X,Y,Z)
plt.contourf(X,Y,Z,cmap="viridis")
plt.title("Contour Plot")
plt.show()
```

```python
# Question-10
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.read_excel("Data/Multiple_Data.xlsx")
x=np.arange(len(df['Category']))
width=0.3
plt.bar(x-width/2,df['Value1'],width,label="Value1")
plt.bar(x+width/2,df['Value2'],width,label="Value2")
plt.xticks(x,df['Category'])
plt.legend()
plt.title("Multiple Bar Chart from Excel")
plt.show()
```
