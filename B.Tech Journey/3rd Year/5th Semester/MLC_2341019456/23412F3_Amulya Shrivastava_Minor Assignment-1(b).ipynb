##### Q1.
- (a) Create a 1D NumPy array with values 1 to 20.  
- (b) Extract all prime numbers from it.  
- (c) Compute the mean and variance of the extracted primes.

```python
import numpy as np

# 1.(a) Create 1D NumPy array with values 1 to 20
arr=np.arange(1,21)
print("Array values: ",arr)

# 1.(b) Extract prime numbers
def is_prime(num):
    if num<2:
        return False
    for i in range(2,int(num**0.5)+1):
        if num%i==0:
            return False
    return True
prime_list=[]
for x in arr:
    if is_prime(x):
        prime_list.append(x)
prime_numbers=np.array(prime_list)
print("Prime numbers are:",prime_numbers)

# 1.(c) Compute mean and variance of primes
print("Mean of the extracted primes: ",prime_numbers.mean())
print("Variance of the extracted primes:",prime_numbers.var())
```

- (b) Extract the 2×2 bottom-left sub-matrix.

```python
# 2.(a) Create 4x4 array with numbers 1 to 16
arr=np.arange(1,17).reshape(4,4)
print("Array of dimensions 4X4: \n",arr)

# 2.(b) Extract 2x2 bottom-left sub-matrix
sub_of_matrix=arr[2:,:2]
print("Bottom-Left Sub-matrix (2X2): \n",sub_of_matrix)

# 2.(c) Compute determinant
determinant=np.linalg.det(sub_of_matrix)
print("Determinant is: ",determinant)
```

##### Q3.
- (a) Create a DataFrame with 5 students and marks in 3 subjects.
- (b) Add a column for total and average marks.
- (c) Identify the topper and print their name with average.

```python
import pandas as pd

# 3.(a) Create DataFrame with 5 students and 3 subjects
data_values={'Name':['Raj','Kaushal','Vikram','Deepti','Akansha'],
            'Math':[85,78,92,70,88],
            'Physics':[80,82,95,65,90],
            'Chemistry':[78,88,91,72,85]}
df=pd.DataFrame(data_values)
print(df,"\n")

# 3.(b) Add Total and Average
df['Total']=df[['Math','Physics','Chemistry']].sum(axis=1)
df['Average']=df['Total']/3
print(df)

# 3.(c) Identify topper
max_avg=df['Average'].max()
topper=df[df['Average']==max_avg].iloc[0]
print(f"\nTopper: {topper['Name']}, Average: {topper['Average']}")
```

##### Q4.
- (a) Simulate 1000 coin tosses using NumPy (1=Head, 0=Tail).
- (b) Count frequency of heads and tails.
- (c) Estimate probability of heads. Is it close to 0.5? Why/Why not?

```python
# 4.(a) Simulate 1000 coin tosses
toss=np.random.randint(0,2,size=1000)

# 4.(b) Count frequency
head=np.sum(toss==1)
tail=np.sum(toss==0)
print("Heads:",head," Tails:",tail)

# 4.(c) Estimate probability of heads
probability_heads=head/1000
print("The Probability of Heads: ",probability_heads)

# Value is close to 0.5 due to law of large numbers.
```

##### Q6.
- (a) Create a 3×3 NumPy array with values 1 to 9.
- (b) Find its transpose and inverse.
- (c) Verify that A × A⁻¹ ≈ I.

```python
# 6.(a) Create 3x3 array with values 1 to 9
Mat_A=np.array([[1,2,3],[0,1,4],[5,6,0]])
print("Matrix A: \n",Mat_A)

# 6.(b) Find transpose and inverse
Mat_A_T=Mat_A.T
print("Transpose: \n",Mat_A_T)

# Inverse
A_inv=np.linalg.inv(Mat_A)
print("Inverse: \n",A_inv)

# 6.(c) Verify A*A_inv=I
Identity=np.dot(Mat_A,A_inv)
print("A*A_inv: \n",Identity)
```

##### Q7.
- (a) Generate random daily temperatures (30 values, range 20–40°C).
- (b) Find the hottest and coldest day.
- (c) Compute mean, median, and standard deviation of temperatures.

```python
# 7.(a) Generate random daily temperatures (30 values,20-40°C)
temperatures=np.random.randint(20,41,30)
print("Temperatures: ",temperatures)

# 7.(b) Hottest and coldest day
hottest_day=0
coldest_day=0
for i in range(1,len(temperatures)):
    if temperatures[i]>temperatures[hottest_day]:
        hottest_day=i
    if temperatures[i]<temperatures[coldest_day]:
        coldest_day=i
print(f"Hottest day: {hottest_day+1}, Temp: {temperatures[hottest_day]}")
print(f"Coldest day: {coldest_day+1}, Temp: {temperatures[coldest_day]}")

# 7.(c) Mean,median,std
print("Mean of temperatures: ",temperatures.mean())
print("Median of temperatures: ",np.median(temperatures))
print("Standard Deviation of temperatures: ",temperatures.std())
```

##### Q8.
- (a) Create a Pandas Series of marks for 8 students.
- (b) Replace all marks below 40 with “Fail”.
- (c) Count how many students passed.

```python
import pandas as pd
import numpy as np

# 8.(a) Create Pandas Series of marks for 8 students
marks=pd.Series([55,38,72,41,29,66,90,35])
print("Marks of students: \n",marks)

# 8.(b) Replace all marks below 40 with "Fail"
marks_replaced=marks.astype(object).copy()
for i in range(len(marks_replaced)):
    if marks_replaced[i]<40:
        marks_replaced[i]="Fail"
print("\nMarks After Replacement: \n",marks_replaced)

# 8.(c) Count how many students passed
pass_marks=(marks>=40).sum()
print("\nNumber of students passed: ",pass_marks)
```

##### Q9.
- (a) Generate 500 random integers between 1 and 6 (simulate dice rolls).
- (b) Count how many times each face appears.
- (c) Compute relative frequencies and compare with theoretical 1/6.

```python
# 9.(a) Simulate 500 dice rolls
dice_roll=np.random.randint(1,7,500)

# 9.(b) Count frequency of each face
counts=[np.count_nonzero(dice_roll==i) for i in range(1,7)]
print("Face counts: ",{i+1:counts[i] for i in range(6)})

# 9.(c) Compute relative frequencies
rel_freq=[counts[i]/500 for i in range(6)]
print("Relative frequencies: ",{i+1:rel_freq[i] for i in range(6)})

# Compare with theoretical probability 1/6 ≈ 0.1667
```

##### Q11.
- (a) Create a DataFrame with some missing values (NaN).
- (b) Fill missing values with column mean.
- (c) Drop rows where more than 1 value is missing.

```python
# 11.(a) Create DataFrame with missing values
df=pd.DataFrame({'A':[1,None,3,4],
                'B':[5,6,np.nan,8],
                'C':[9,10,11,None]})
print("Original DataFrame:\n",df)

# 11.(b) Fill missing values with column mean
df1=df.fillna(df.mean())
print("\nAfter filling NaN with column mean: \n",df1)

# 11.(c) Drop rows where more than 1 value is missing
check=df1.isnull().sum(axis=1)<=1
df2=df1[check]
print("\nAfter dropping rows with >1 missing: \n",df2)
```

##### Q14.
- (a) Create a Pandas DataFrame with 5 employees having columns: Name, Tasks_Completed, Hours_Worked.
- (b) Add a new column Efficiency = Tasks_Completed / Hours_Worked.
- (c) Identify the employee with the highest efficiency and print their name and value.

```python
# 14.(a) Create DataFrame with 5 employees
employees=pd.DataFrame({'Name':['Raj','Kaushal','Vikram','Deepti','Akansha'],
                        'Tasks_Completed':[50,60,45,70,55],
                        'Hours_Worked':[40,50,35,60,45]})
print(employees,"\n")

# 14.(b) Add Efficiency = Tasks_Completed/Hours_Worked
employees['Efficiency']=employees['Tasks_Completed']/employees['Hours_Worked']
print(employees)

# 14.(c) Employee with highest efficiency
max_eff=employees['Efficiency'].max()
best_emp=employees[employees['Efficiency']==max_eff].iloc[0]
print(f"\nHighest Efficiency: {best_emp['Name']}, Value: {best_emp['Efficiency']}")
```
