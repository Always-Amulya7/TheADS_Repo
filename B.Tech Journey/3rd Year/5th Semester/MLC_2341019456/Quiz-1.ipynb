```python
#Fibonacci nos generation using lists
fibo=[0,1]
n=int(input("Enter the number: "))
for i in range(2,n):
    fibo.append(fibo[-1]+fibo[-2])
    print(fibo)
```

```python
# Star pattern
num=int(input("Enter the length: "))
for i in range(1,num):
    for j in range(1,i+1):
        print("*",end=" ")
    print()
```

```python
# Palindrome Check
enter=input("Enter the String to be checked: ")
rev=enter[::-1]
if (enter==rev):
    print("It's Palindrome")
else:
    print("Not a Palindrome")
```

```python
# Factorial Number Generation
num=int(input("Enter the number: "))
fact=1
for i in range(1,num+1):
    fact*=i
print(f"The factorial of {num} is",fact)
```

```python
# Inverse ope
import numpy as np
matrix=np.arange(1, 10).reshape(3,3)
det=np.linalg.det(matrix)
inverse=np.linalg.inv(matrix)
print("The Determinant is: ",det)
print("The Inverse is: ",inverse)
identity_matrix=np.eye(3)
check=np.allclose(np.dot(matrix,inverse),identity_matrix)
if check:
    print("The inverse exists")
```

```python
import numpy as np
import pandas as pd
df=pd.DataFrame({'Price':[20_000,30_000,35_000,80_000]})
# inf is infinity
bins=[0,10000,60000,np.inf]
label=["low","medium","high"]
# Cut - write inclusive, automatically arrange with high,med and low
df["Price_Category"]=pd.cut(df["Price"],bins=bins,labels=label)
df
```

```python
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
iris=load_iris()
print(iris)
X=iris.data
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X) 
print("Original shape:",X.shape)
print("Reduced shape :",X_pca.shape)
```
