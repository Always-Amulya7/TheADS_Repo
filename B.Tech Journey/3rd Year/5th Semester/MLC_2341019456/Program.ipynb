```python
# CLT demonstration

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

pop = np.random.exponential(scale=1.0, size=200000)  # Population
sample_size = 50
n_trials = 2000
means = [np.mean(np.random.choice(pop, sample_size)) for _ in range(n_trials)]
plt.figure(figsize=(6, 4))
plt.hist(means, bins=40, density=True)

# Overlay normal with same mean and variance of sampling distribution
mu_hat = np.mean(means)
sigma_hat = np.std(means)   # Empirical std of the sampling distribution
x = np.linspace(mu_hat - 4 * sigma_hat, mu_hat + 4 * sigma_hat, 200)
plt.plot(x, stats.norm.pdf(x, mu_hat, sigma_hat), 'r')
plt.title(f'CLT: distribution of sample means (n={sample_size})')
plt.show()
print("Sample means mean:", mu_hat, "std:", sigma_hat)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
print("\n---K-Nearest Example---")
X=np.array([[1,2],[2,3],[3,3],[6,5],[7,7],[8,6]])
y=np.array([0,0,0,1,1,1])
k=3
knn=KNeighborsClassifier(n_neighbors=k)
knn.fit(X,y)
sample=np.array([[3,5]])
predicted_class=knn.predict(sample)[0]
print("Predicted class for [3,5]:", predicted_class)
h=0.1
x_min,x_max=X[:,0].min()-1,X[:,0].max()+1
y_min,y_max=X[:,1].min()-1,X[:,1].max()+1
xx,yy=np.meshgrid(np.arange(x_min,x_max,h),np.arange(y_min,y_max,h))
Z=knn.predict(np.c_[xx.ravel(),yy.ravel()])
Z=Z.reshape(xx.shape)
plt.contourf(xx,yy,Z,alpha=0.3,cmap=plt.cm.Paired)
plt.scatter(X[:,0],X[:,1],c=y,edgecolor='k',s=80,cmap=plt.cm.Paired,label="Training Points")
plt.scatter(sample[0,0],sample[0,1],c='red',edgecolor='black',s=120, marker="*",label="Test Sample [3,5]")
plt.text(1,1.5,"Class 0 region",fontsize=10,color='blue')
plt.text(6.2,4.2,"Class 1 region",fontsize=10,color='darkgreen')
plt.title(f"KNN Decision Boundary (k={k})", fontsize=13)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# DataSets
X=np.array([15,23,18,23,24,22,22,19,19,18,24,11,24,16,23]).reshape(-1,1)
Y=np.array([49,63,58,60,58,61,60,63,60,52,52,30,59,49,60])

# Fit Model
lr=LinearRegression().fit(X,Y)

plt.scatter(X,Y)
plt.plot(X,lr.predict(X))
plt.xlabel("Internal Marks")
plt.ylabel("External Marks")
plt.title("Linear Regression")
plt.show()

print("Intercept: ", lr.intercept_)
print("Slope: ",lr.coef_[0])
print("A^2 Score: ",lr.score(X,Y))
```
