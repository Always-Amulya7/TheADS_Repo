```python
def power(base, exponent):
    if exponent == 1:
        return base
    return base * power(base, exponent - 1)
base = int(input())
exponent = int(input())
print("The exponent solution is: ",power(base, exponent))
```

```python
def gcd(x, y):
    if y == 0:
        return x
    return gcd(y, x % y)
x = int(input())
y = int(input())
print("The GCD is: ",gcd(x, y))
```

```python
def generate(current, last, n):
    if len(current) == n:
        print("".join(map(str, current)))
        return
    for i in range(last + 1, 10):
        generate(current + [i], i, n)
n = int(input())
generate([], -1, n)
```

```python
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)
def fib_iterative(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
n = int(input())
print("The Fibonacci By Recursion Is: ",fib_recursive(n))
print("O(2^n)")
print("The Fibonacci By Iteration Is: ",fib_iterative(n))
print("O(n)")
```

```python
import random
def quickselect(arr, k):
    pivot = random.choice(arr)
    left = [x for x in arr if x > pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x < pivot]
    if k <= len(left):
        return quickselect(left, k)
    if k <= len(left) + len(mid):
        return pivot
    return quickselect(right, k - len(left) - len(mid))
arr = list(map(int, input().split()))
k = int(input())
print("Found At: ",quickselect(arr, k))
```

```python
import math
def has_antipodal(points):
    angles = []
    for x, y in points:
        a = math.atan2(y, x)
        if a < 0:
            a += 2 * math.pi
        angles.append(round(a, 10))
    angles.sort()
    s = set(angles)
    for a in angles:
        b = round((a + math.pi) % (2 * math.pi), 10)
        if b in s:
            return True
    return False
n = int(input())
points = [tuple(map(float, input().split())) for _ in range(n)]
print("Does this has antipodal?: ",has_antipodal(points))
```

```python
def partition(arr, low, high):
    pivot = arr[low]
    i = low + 1
    j = high
    while True:
        while i <= j and arr[i] <= pivot:
            i += 1
        while i <= j and arr[j] > pivot:
            j -= 1
        if i <= j:
            arr[i], arr[j] = arr[j], arr[i]
        else:
            break
    arr[low], arr[j] = arr[j], arr[low]
    return j
def quick_sort_helper(arr, low, high):
    if low < high:
        p = partition(arr, low, high)
        quick_sort_helper(arr, low, p - 1)
        quick_sort_helper(arr, p + 1, high)
def quick_sort(arr):
    quick_sort_helper(arr, 0, len(arr) - 1)
arr = [37, 2, 6, 4, 89, 8, 10, 12, 68, 45]
quick_sort(arr)
print("The Sorted Array Is: ",arr)
```

```python
data = [
    ("Elon Musk", 433.9),
    ("Jeff Bezos", 239.4),
    ("Mark Zuckerberg", 211.8),
    ("Larry Ellison", 204.6),
    ("Bernard Arnault & Family", 181.3),
    ("Larry Page", 161.4)
]
def selection_sort(arr):
    a = arr[:]
    n = len(a)
    for i in range(n):
        m = i
        for j in range(i + 1, n):
            if a[j][1] < a[m][1]:
                m = j
        a[i], a[m] = a[m], a[i]
    return dict(a)
def bubble_sort(arr):
    a = arr[:]
    n = len(a)
    for i in range(n):
        for j in range(0, n - i - 1):
            if a[j][1] > a[j + 1][1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return dict(a)
def insertion_sort(arr):
    a = arr[:]
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j][1] > key[1]:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return dict(a)
print("By Selection Sort: ")
print(selection_sort(data))
print("By Bubble Sort: ")
print(bubble_sort(data))
print("By Insertion Sort: ")
print(insertion_sort(data))
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
arr = input().split()
print("By Merge Sort: ")
print(merge_sort(arr))
```

```python
def merge_lists(a, b):
    i = j = 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    result.extend(a[i:])
    result.extend(b[j:])
    return result
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print("By Merge Sort And Merge List: ")
print(merge_lists(a, b))
```
