```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
text = """Natural Language Processing enables machines to understand and process human languages.
It is a fascinating field with numerous applications, such as chatbots and language translation."""
words = word_tokenize(text)
sentences = sent_tokenize(text)
ps = PorterStemmer()
lm = WordNetLemmatizer()
stemmed = [ps.stem(word) for word in words]
lemmatized = [lm.lemmatize(word) for word in words]
stop_words = set(stopwords.words("english"))
filtered = [word for word in words if word.lower() not in stop_words and word.isalpha()]
print("TextBlob Words")
print(words)
print("TextBlob Sentences")
print(sentences)
print("TextBlob Stem")
print(stemmed)
print("TextBlob Lemma")
print(lemmatized)
print("TextBlob Filtered")
print(filtered)
```

```python
import requests
from bs4 import BeautifulSoup
from textblob import TextBlob
from textblob.sentiments import NaiveBayesAnalyzer
html = requests.get("https://www.bbc.com/news").text
soup = BeautifulSoup(html, "html.parser")
text = soup.get_text()
blob = TextBlob(text, analyzer=NaiveBayesAnalyzer())
print("The Sentiment Analyzed: ",blob.sentiment)
for sentence in blob.sentences:
    s = TextBlob(str(sentence), analyzer=NaiveBayesAnalyzer())
    print(sentence, s.sentiment,"\n")
```

```python
from textblob import TextBlob
from deep_translator import GoogleTranslator
from collections import Counter
text = input()
blob = TextBlob(text)
print("In French")
print(GoogleTranslator(source="auto", target="fr").translate(text))
print("In Spanish")
print(GoogleTranslator(source="auto", target="es").translate(text))
print("In German")
print(GoogleTranslator(source="auto", target="de").translate(text))
print(blob.sentiment)
for s in blob.sentences:
    print(s, s.sentiment)
print("TextBlob Tags: ")
print(blob.tags)
word = input()
w = TextBlob(word)
print(w.correct())
adjectives = [word for word, tag in blob.tags if tag.startswith("JJ")]
print("Adjectives: ",adjectives)
np = Counter(blob.noun_phrases)
print("Commons: ",np.most_common(5))
```

```python
from nltk.corpus import wordnet
word = input()
synsets = wordnet.synsets(word)
if synsets:
    print(synsets[0].definition())
synonyms = set()
antonyms = set()
for syn in synsets:
    for lemma in syn.lemmas():
        synonyms.add(lemma.name())
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())
print("The Synonyms")
print(list(synonyms))
print("The Antonyms")
print(list(antonyms))
```

```python
import requests
from bs4 import BeautifulSoup
from textatistic import Textatistic
import re
url = input()
html = requests.get(url).text
soup = BeautifulSoup(html, "html.parser")
text = soup.get_text(separator=" ")
text = re.sub(r"\s+", " ", text)
sentences = re.split(r"[.!?]+", text)
clean_sentences = []
for s in sentences:
    words = [w for w in s.split() if w.isalpha() and len(w) <= 20]
    if len(words) > 2:
        clean_sentences.append(" ".join(words))
clean_text = ". ".join(clean_sentences)
stats = Textatistic(clean_text)
avg_word_per_sent = stats.word_count / stats.sent_count
avg_character_per_word = stats.char_count / stats.word_count
avg_syllables_per_word = stats.sybl_count / stats.word_count
print("Word Count: ",stats.word_count)
print("Word Sent Count: ",stats.sent_count)
print("Average Word Per Sent: ",avg_word_per_sent)
print("Average Charactr Per Word: ",avg_character_per_word)
print("Average Syllables Per Word",avg_syllables_per_word)
```

```python
import requests
from bs4 import BeautifulSoup
import spacy
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")
url = input()
html = requests.get(url).text
soup = BeautifulSoup(html, "html.parser")
text = soup.get_text(separator=" ")
doc = nlp(text)
print("Token Analysis: ")
for token in doc:
    if token.text.istitle() and token.is_alpha:
        print(token.text, "ENTITY")
```

```python
import requests
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
text1 = requests.get("https://www.gutenberg.org/files/1513/1513-0.txt").text
text2 = requests.get("https://www.gutenberg.org/files/1112/1112-0.txt").text
docs = [text1[:50000], text2[:50000]]
vectorizer = TfidfVectorizer(stop_words="english")
tfidf = vectorizer.fit_transform(docs)
score = cosine_similarity(tfidf[0:1], tfidf[1:2])
print("The Numerical Score: ",score[0][0])
```
