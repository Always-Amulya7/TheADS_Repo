```python
class BankAccount:
    def __init__(self, account_number, account_holder, initial_balance=0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = initial_balance
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            return f"Deposited ${amount}. New balance: ${self.balance}"
        return "Invalid deposit amount"
    
    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            return f"Withdrawn ${amount}. New balance: ${self.balance}"
        return "Insufficient funds or invalid amount"
    
    def display_balance(self):
        return f"Account {self.account_number} - Balance: ${self.balance}"

account = BankAccount("123456", "John Doe", 1000)
print(account.deposit(500))
print(account.withdraw(200))
print(account.display_balance())
```

```python
class Chapter:
    def __init__(self, title, page_count):
        self.title = title
        self.page_count = page_count

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.chapters = []
    
    def add_chapter(self, chapter):
        self.chapters.append(chapter)
    
    def get_total_pages(self):
        return sum(chapter.page_count for chapter in self.chapters)
    
    def display_book_info(self):
        print(f"Book: {self.title} by {self.author}")
        print("Chapters:")
        for i, chapter in enumerate(self.chapters, 1):
            print(f"  {i}. {chapter.title} - {chapter.page_count} pages")
        print(f"Total pages: {self.get_total_pages()}")

# Create book with three chapters
book = Book("Python Programming", "Jane Smith")
book.add_chapter(Chapter("Introduction", 20))
book.add_chapter(Chapter("Data Types", 35))
book.add_chapter(Chapter("Object-Oriented Programming", 45))
book.display_book_info()
```

```python
class Time:
    def __init__(self, time_string):
        self.time_string = time_string
        self.hours = 0
        self.minutes = 0
        self.seconds = 0
    
    def validate_time(self):
        """Validate if time string is in HH:MM:SS format"""
        try:
            parts = self.time_string.split(':')
            if len(parts) != 3:
                return False, "Time must be in HH:MM:SS format"
            
            self.hours = int(parts[0])
            self.minutes = int(parts[1])
            self.seconds = int(parts[2])
            
            if not (0 <= self.hours <= 23):
                return False, "Hours must be between 0 and 23"
            if not (0 <= self.minutes <= 59):
                return False, "Minutes must be between 0 and 59"
            if not (0 <= self.seconds <= 59):
                return False, "Seconds must be between 0 and 59"
            
            return True, "Valid time"
        except ValueError:
            return False, "Time components must be integers"
    
    def convert_to_12hour(self):
        """Convert 24-hour format to 12-hour format with AM/PM"""
        is_valid, message = self.validate_time()
        if not is_valid:
            return message
        
        period = "AM"
        hour_12 = self.hours
        
        if self.hours == 0:
            hour_12 = 12
        elif self.hours == 12:
            period = "PM"
        elif self.hours > 12:
            hour_12 = self.hours - 12
            period = "PM"
        
        return f"{hour_12:02d}:{self.minutes:02d}:{self.seconds:02d} {period}"

# Test the Time class
times = ["14:30:45", "09:15:30", "25:00:00", "12:00:00", "00:30:15"]
for time_str in times:
    t = Time(time_str)
    result = t.convert_to_12hour()
    print(f"{time_str} -> {result}")
```

```python
import random

class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
    
    def __str__(self):
        return f"{self.rank} of {self.suit}"

class Deck:
    suits = ['Hearts', 'Diamonds', 'Clubs', 'Spades']
    ranks = ['Ace', '2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King']
    
    def __init__(self):
        self.cards = []
        self.reset()
    
    def reset(self):
        self.cards = [Card(suit, rank) for suit in self.suits for rank in self.ranks]
    
    def shuffle(self):
        random.shuffle(self.cards)
        return "Deck shuffled"
    
    def deal_card(self):
        if len(self.cards) > 0:
            return self.cards.pop()
        return None

class Player:
    def __init__(self, name):
        self.name = name
        self.hand = []
    
    def receive_card(self, card):
        if card:
            self.hand.append(card)
    
    def show_hand(self):
        if not self.hand:
            return f"{self.name}'s hand is empty"
        hand_str = f"{self.name}'s hand:\n"
        for i, card in enumerate(self.hand, 1):
            hand_str += f"  {i}. {card}\n"
        return hand_str

# Simulate the game
deck = Deck()
deck.shuffle()

player1 = Player("Alice")
player2 = Player("Bob")

# Deal 5 cards to each player
for _ in range(5):
    player1.receive_card(deck.deal_card())
    player2.receive_card(deck.deal_card())

print(player1.show_hand())
print(player2.show_hand())
print(f"Remaining cards in deck: {len(deck.cards)}")
```

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model
    
    def display_info(self):
        return f"Vehicle: {self.make} {self.model}"

class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)  # Call parent class constructor
        self.num_doors = num_doors
    
    def display_info(self):
        # Extend the parent class functionality
        return f"Car: {self.make} {self.model} with {self.num_doors} doors"

# Instantiate objects
vehicle = Vehicle("Generic", "Model X")
car = Car("Toyota", "Camry", 4)

print(vehicle.display_info())
print(car.display_info())

"""
Explanation of inheritance:
- The Car class inherits all attributes and methods from Vehicle
- It adds its own attribute (num_doors)
- It overrides display_info() to provide more specific information
- super() is used to call the parent class constructor
- The subclass extends functionality by adding new features while reusing existing code
"""
```

```python
import math

class Shape:
    def area(self):
        pass  # Base method to be overridden

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

# Polymorphism in action
shapes = [Circle(5), Rectangle(4, 6)]

for shape in shapes:
    print(f"Area: {shape.area():.2f}")

"""
How polymorphism simplifies working with different shapes:
1. Same interface (area() method) works for all shapes
2. No need to check the type of object before calling area()
3. New shapes can be added without changing existing code
4. Code becomes more flexible and maintainable
5. Follows the "open/closed principle" - open for extension, closed for modification
"""
```

```python
class CommissionEmployee:
    def __init__(self, first_name, last_name, ssn, gross_sales, commission_rate):
        self.first_name = first_name
        self.last_name = last_name
        self.ssn = ssn
        self.gross_sales = gross_sales
        self.commission_rate = commission_rate
    
    @property
    def gross_sales(self):
        return self._gross_sales
    
    @gross_sales.setter
    def gross_sales(self, value):
        if value < 0:
            raise ValueError("Gross sales cannot be negative")
        self._gross_sales = value
    
    @property
    def commission_rate(self):
        return self._commission_rate
    
    @commission_rate.setter
    def commission_rate(self, value):
        if value < 0 or value > 1:
            raise ValueError("Commission rate must be between 0 and 1")
        self._commission_rate = value
    
    def earnings(self):
        return self.gross_sales * self.commission_rate
    
    def __repr__(self):
        return (f"CommissionEmployee({self.first_name}, {self.last_name}, "
                f"{self.ssn}, Gross Sales: ${self.gross_sales:.2f}, "
                f"Rate: {self.commission_rate:.2%})")

# Test script
try:
    # Instantiate object
    emp = CommissionEmployee("John", "Doe", "123-45-6789", 50000, 0.15)
    print(emp)
    print(f"Earnings: ${emp.earnings():.2f}")
    
    # Modify sales data
    emp.gross_sales = 75000
    print("\nAfter updating gross sales:")
    print(emp)
    print(f"Earnings: ${emp.earnings():.2f}")
    
    # Test validation
    print("\nTesting validation (expect error):")
    emp.gross_sales = -1000  # This will raise ValueError
    
except ValueError as e:
    print(f"Error: {e}")
```

```python
"""
Duck typing: "If it walks like a duck and quacks like a duck, it's a duck"
Python focuses on what an object can do, not what it is.
"""

class Dog:
    def speak(self):
        return "Woof!"

class Robot:
    def speak(self):
        return "Beep boop!"

class Cat:
    def meow(self):
        return "Meow!"

def describe(obj):
    """Function accepts any object with a speak() method"""
    print(f"The object says: {obj.speak()}")

# Duck typing in action
dog = Dog()
robot = Robot()
cat = Cat()

describe(dog)    # Works - Dog has speak()
describe(robot)  # Works - Robot has speak()
# describe(cat)  # AttributeError - Cat doesn't have speak()

"""
Explanation:
- describe() doesn't check the type of object passed
- It only cares that the object has a speak() method
- This makes the code flexible and reusable
- New classes can be used without modifying existing code
"""
```

```python
class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    
    def __add__(self, other):
        """Overload the + operator"""
        return Complex(self.real + other.real, self.imag + other.imag)
    
    def __str__(self):
        sign = '+' if self.imag >= 0 else ''
        return f"{self.real}{sign}{self.imag}i"

# Test the class
c1 = Complex(3, 4)
c2 = Complex(1, 2)
c3 = c1 + c2

print(f"c1 = {c1}")
print(f"c2 = {c2}")
print(f"c1 + c2 = {c3}")
```

```python
class InsufficientFundsError(Exception):
    """Custom exception for insufficient funds"""
    def __init__(self, balance, withdrawal_amount):
        self.balance = balance
        self.withdrawal_amount = withdrawal_amount
        self.message = f"Insufficient funds! Balance: ${balance}, Withdrawal attempt: ${withdrawal_amount}"
        super().__init__(self.message)

class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return f"Withdrawn ${amount}. New balance: ${self.balance}"

# Test the custom exception
account = BankAccount(100)

try:
    print(account.withdraw(50))   # Should work
    print(account.withdraw(100))  # Should raise exception
except InsufficientFundsError as e:
    print(f"Error: {e}")
```

```python
from dataclasses import dataclass
import random

@dataclass
class Card:
    suit: str
    rank: str
    
    def __str__(self):
        return f"{self.rank} of {self.suit}"

class Deck:
    suits = ['Hearts', 'Diamonds', 'Clubs', 'Spades']
    ranks = ['Ace', '2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King']
    
    def __init__(self):
        self.cards = [Card(suit, rank) for suit in self.suits for rank in self.ranks]
        self.shuffle()
    
    def shuffle(self):
        random.shuffle(self.cards)
    
    def deal_cards(self, num_cards):
        dealt = []
        for _ in range(min(num_cards, len(self.cards))):
            dealt.append(self.cards.pop())
        return dealt

# Simulate dealing 5 cards
deck = Deck()
player_hand = deck.deal_cards(5)

print("Player's hand:")
for i, card in enumerate(player_hand, 1):
    print(f"  {i}. {card}")

print(f"\nRemaining cards in deck: {len(deck.cards)}")
```

```python
from dataclasses import dataclass
from typing import NamedTuple
from collections import namedtuple

# Named Tuple example
PointNT = namedtuple('PointNT', ['x', 'y'])
p1 = PointNT(10, 20)

# Data Class example
@dataclass
class PointDC:
    x: float
    y: float
    
    def distance_from_origin(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5
    
    def move(self, dx, dy):
        self.x += dx
        self.y += dy

# Test both
p_dc = PointDC(3, 4)
print(f"Data Class distance: {p_dc.distance_from_origin()}")
p_dc.move(1, 1)
print(f"After move: ({p_dc.x}, {p_dc.y})")

"""
Advantages of Data Classes over Named Tuples:
1. Mutable by default (can change values)
2. Can have methods
3. Default values for fields
4. Inheritance support
5. More flexibility with field types
6. Can use type hints properly
"""
```

```python
species = "Global Species"

class Animal:
    species = "Class Species"
    
    def __init__(self, species):
        self.species = species
    
    def display_species(self):
        print("Instance species:", self.species)
        print("Class species:", Animal.species)
        print("Global species:", globals()['species'])

# Create instance and call method
a = Animal("Instance Species")
a.display_species()

"""
OUTPUT:
Instance species: Instance Species
Class species: Class Species
Global species: Global Species

SCOPE RESOLUTION PROCESS:
1. Instance species (self.species): Looks in the object's namespace first
   → Found "Instance Species" (passed to constructor)

2. Class species (Animal.species): Looks in the class namespace
   → Found "Class Species" (class attribute)

3. Global species (globals()['species']): Looks in global namespace
   → Found "Global Species" (global variable)

4. Built-in namespace would be checked if not found in previous levels
"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create Celsius to Kelvin conversion lambda
celsius_to_kelvin = lambda c: c + 273.15

# Generate Celsius temperatures
celsius_temps = np.arange(0, 101, 10)

# Convert to Kelvin using lambda
kelvin_temps = [celsius_to_kelvin(c) for c in celsius_temps]

# Create DataFrame in tabular format
df = pd.DataFrame({
    'Celsius (°C)': celsius_temps,
    'Kelvin (K)': kelvin_temps
})

# Display the table
print("Temperature Conversion Table:")
print(df.to_string(index=False))

# Visualize the data
plt.figure(figsize=(10, 6))
plt.plot(df['Celsius (°C)'], df['Kelvin (K)'], 'bo-', linewidth=2, markersize=8)
plt.xlabel('Celsius (°C)')
plt.ylabel('Kelvin (K)')
plt.title('Celsius to Kelvin Temperature Conversion')
plt.grid(True, alpha=0.3)

# Add data labels
for i, (c, k) in enumerate(zip(celsius_temps, kelvin_temps)):
    plt.annotate(f'{k:.1f}K', (c, k), textcoords="offset points", 
                 xytext=(0,10), ha='center')

plt.show()
```
