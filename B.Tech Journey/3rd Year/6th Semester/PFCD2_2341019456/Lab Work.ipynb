```python
from decimal import Decimal
class Account:
    "Bank Account"
    def __init__(self,name,balance):
        if balance<Decimal('0.00'):
            raise ValueError("Initial Balance Must Be >= 0.0")
        self.name=name
        self.balance=balance
    def deposit(self,amount):
        if amount<Decimal('0.00'):
            raise ValueError("Amount Must Be Positive")
        self.balance+=amount
    def withdraw(self,amount):
        if amount>self.balance:
            raise ValueError("Amount Must Be <= Balance")
        elif amount<Decimal('0.00'):
            raise ValueError('Amount Must Be Positive')
        self.balance-=amount
```

```python
from decimal import Decimal
acc1=Account('John',Decimal('50.00'))
acc1.withdraw(Decimal('20.00'))
print(acc1.balance)
acc1.deposit(Decimal('500.00'))
print(acc1.balance)
```

```python
class Time:
    '''Time class to represent time with hour, minute, and second'''
    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second

    @property
    def hour(self):
        return self._hour

    @hour.setter
    def hour(self, hour):
        if not (0 <= hour < 24):
            raise ValueError("Hour must be between 0 and 23")
        self._hour = hour

    @property
    def minute(self):
        return self._minute

    @minute.setter
    def minute(self, minute):
        if not (0 <= minute < 60):
            raise ValueError("Minute must be between 0 and 59")
        self._minute = minute

    @property
    def second(self):
        return self._second

    @second.setter
    def second(self, second):
        if not (0 <= second < 60):
            raise ValueError("Second must be between 0 and 59")
        self._second = second

    def set_time(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def display(self):
        print(f"{self.hour:02d}:{self.minute:02d}:{self.second:02d}")
```

```python
# Create object
t1 = Time(10, 30, 45)

# Display time
print("Initial Time:")
t1.display()

# Change time using set_time method
t1.set_time(15, 20, 10)

# Display updated time
print("\nUpdated Time:")
t1.display()

# Access properties individually
print("\nIndividual Components:")
print(f"Hour: {t1.hour}")
print(f"Minute: {t1.minute}")
print(f"Second: {t1.second}")
```

Stemming and Lemmatization

Stemming removes a prefix or suffix from the word living only stem, which may or not be real word.

Lemmatization is similar but factors in words part of speech and meaning and result and in real word.

Stemming and Lemmatization are normalization operations, in which you prepare words for analysis.
