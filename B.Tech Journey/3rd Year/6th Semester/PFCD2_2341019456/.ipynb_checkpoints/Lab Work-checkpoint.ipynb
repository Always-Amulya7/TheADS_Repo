```python
class Student:
    def __init__(Self):
        Self.name='Amulya'
        Self.age=21
        Self.marks=95
    def talk(Self):
        print("My name is",Self.name)
        print("My age is",Self.age)
        print("My marks is",Self.marks)
s=Student()
s.talk()
```

```python
class Student:
    def __init__(Self,name,age,marks):
        Self.name=name
        Self.age=age
        Self.marks=marks
    def talk(Self):
        print("My name is",Self.name)
        print("My age is",Self.age)
        print("My marks is",Self.marks)
s=Student("Krrish",14,89)
s.talk()
```
