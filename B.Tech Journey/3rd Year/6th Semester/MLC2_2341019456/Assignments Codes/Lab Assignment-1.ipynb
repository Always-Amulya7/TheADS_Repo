### Dataset and Visualization 

1. **Create a small synthetic dataset** containing at least 10 two-dimensional points. Use NumPy to store the data. 
2. **Visualize the dataset** using a scatter plot. Label the axes and give the plot an appropriate title. 

```python
import numpy as np
data = np.array([[2,3],[3,4],[3,3],[5,6],[8,7],[7,5],[6,8],[1,2],[9,8],[8,9]])
print(data)
```

```python
import matplotlib.pyplot as plt
plt.scatter(data[:,0],data[:,1])
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Synthetic Dataset")
plt.show()
```

### Elbow Method 

3. **Implement the Elbow Method for K-Means clustering** on your dataset. Compute the Within-Cluster Sum of Squares (WCSS) for K values ranging from 1 to 5. 
4. **Plot the WCSS values** against K and label the axes. Explain how you would use the plot to select the optimal number of clusters.

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,6):
    kmeans=KMeans(n_clusters=i,random_state=42)
    kmeans.fit(data)
    wcss.append(kmeans.inertia_)
print(wcss)
```

```python
plt.plot(range(1,6),wcss,marker='o')
plt.xlabel("K")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

### K-Means Clustering

5. **Perform K-Means clustering** on your dataset using 3 clusters. Assign cluster labels to each data point. 
6. **Plot the K-Means clusters** on a scatter plot and show the cluster centroids. Use different colors for each cluster.

```python
kmeans=KMeans(n_clusters=3,random_state=42)
labels=kmeans.fit_predict(data)
centroids=kmeans.cluster_centers_
print(labels)
print(centroids)
```

```python
plt.scatter(data[:,0],data[:,1],c=labels)
plt.scatter(centroids[:,0],centroids[:,1],marker='X',s=200)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("K-Means Clustering")
plt.show()
```

### Hierarchical Clustering 

7. **Compute hierarchical clustering linkage** for your dataset using the Ward method. 
8. **Plot a dendrogram** to visualize hierarchical clustering. Label the axes and provide a title. 
9. **Assign clusters from the dendrogram** by cutting it to form 3 clusters. Visualize the clusters in a scatter plot.

```python
from scipy.cluster.hierarchy import linkage
linkage_matrix=linkage(data,method='ward')
print(linkage_matrix)
```

```python
from scipy.cluster.hierarchy import dendrogram
dendrogram(linkage_matrix)
plt.title("Hierarchical Dendrogram")
plt.xlabel("Points")
plt.ylabel("Distance")
plt.show()
```

```python
from scipy.cluster.hierarchy import fcluster
h_labels=fcluster(linkage_matrix,3,criterion='maxclust')
plt.scatter(data[:,0],data[:,1],c=h_labels)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Hierarchical Clustering")
plt.show()
```

### DBSCAN Clustering 

10. **Perform DBSCAN clustering** on your dataset. Choose appropriate values for eps and min_samples. 
11. **Visualize DBSCAN clusters** on a scatter plot. Make sure to indicate noise points if any.

```python
from sklearn.cluster import DBSCAN
dbscan=DBSCAN(eps=2,min_samples=2)
db_labels=dbscan.fit_predict(data)
print(db_labels)
```

```python
plt.scatter(data[:,0],data[:,1],c=db_labels)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("DBSCAN Clustering")
plt.show()
```
