#### Biological Neuron Simulation

#### Question-1
Write a Python program that simulates a biological neuron firing mechanism. The neuron should fire only if the sum of signals exceeds a given threshold, and modify the neuron simulation so that the threshold value is taken from the user.

```python
threshold=int(input("Enter threshold: "))
signals=[1,0,1,1]
s=sum(signals)
if s>=threshold:
    print("Neuron Fired")
else:
    print("Neuron Not Fired")
```

#### Question-5
Write a Python function that accepts an activation function name as input and returns the corresponding output value. 

```python
import math
def activation(name,x):
    if name=="identity":
        return x
    if name=="relu":
        if x>0:
            return x
        else:
            return 0
    if name=="sigmoid":
        return 1/(1+math.exp(-x))
    if name=="tanh":
        return math.tanh(x)

name=input("Enter function: ")
x=float(input("Enter value: "))
print(activation(name,x))
```

#### McCulloch-Pitts

#### Question-6
Write a Python program that implements a McCulloch–Pitts neuron to simulate the NAND gate. 

```python
def nand(x1,x2):
    w1=-2
    w2=-2
    b=3
    s=x1*w1+x2*w2+b
    if s>=0:
        return 1
    else:
        return 0

inputs=[(0,0),(0,1),(1,0),(1,1)]
for i in inputs:
    print(i,nand(i[0],i[1]))
```

#### Question-7
Create a program that prints the truth table for XOR logic using two neurons.

```python
def neuron(x1,x2,w1,w2,b):
    s=x1*w1+x2*w2+b
    if s>=0:
        return 1
    else:
        return 0

inputs=[(0,0),(0,1),(1,0),(1,1)]
print("x1 x2 xor")
for x1,x2 in inputs:
    n1=neuron(x1,x2,1,1,-1)
    n2=neuron(x1,x2,1,1,-2)
    xor=neuron(n1,n2,1,-2,0)
    print(x1,x2,xor)
```

#### Perceptron Model

#### Question-8
Write a Python program to create a perceptron classifier and train it using a small dataset with two features. 

```python
X=[[0,0],[0,1],[1,0],[1,1]]
y=[0,0,0,1]
w=[0,0]
b=0
lr=0.1
for epoch in range(10):
    for i in range(len(X)):
        s=X[i][0]*w[0]+X[i][1]*w[1]+b
        if s>=0:
            y_pred=1
        else:
            y_pred=0
        e=y[i]-y_pred
        w[0]+=lr*e*X[i][0]
        w[1]+=lr*e*X[i][1]
        b+=lr*e
print(w,b)
```

#### Question-11
Write a Python program that demonstrates the effect of different learning rates on perceptron training. 

```python
X=[[0,0],[0,1],[1,0],[1,1]]
y=[0,0,0,1]
rates=[0.1,0.5,1]
for lr in rates:
    w=[0,0]
    b=0
    for epoch in range(5):
        for i in range(len(X)):
            s=X[i][0]*w[0]+X[i][1]*w[1]+b
            if s>=0:
                y_pred=1
            else:
                y_pred=0
            e=y[i]-y_pred
            w[0]+=lr*e*X[i][0]
            w[1]+=lr*e*X[i][1]
            b+=lr*e
    print("LR",lr,"Weights",w,"Bias",b)
```

#### ADALINE Model

#### Question-13
Write a Python program to implement ADALINE learning using the delta rule for a simple dataset and modify the ADALINE program to print the training error after each epoch.

```python
X=[[1,1],[2,1],[3,2],[4,3]]
y=[2,3,4,5]
w=[0,0]
b=0
lr=0.01
for epoch in range(10):
    error=0
    for i in range(len(X)):
        y_pred=w[0]*X[i][0]+w[1]*X[i][1]+b
        e=y[i]-y_pred
        w[0]+=lr*e*X[i][0]
        w[1]+=lr*e*X[i][1]
        b+=lr*e
        error+=e*e
    print("Epoch",epoch,"Error",error)
```
