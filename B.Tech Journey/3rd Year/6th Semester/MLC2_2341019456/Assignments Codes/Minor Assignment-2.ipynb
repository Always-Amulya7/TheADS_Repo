```python
# 3 (a)
import numpy as np
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

X, y = make_classification(n_samples=120, n_features=2, n_redundant=0, n_informative=2, n_clusters_per_class=1, random_state=42)

labeled_idx = list(range(10))
unlabeled_idx = list(range(10, 120))


# 3 (b) (c) (d) (e)
for _ in range(8):
    model = LogisticRegression()
    model.fit(X[labeled_idx], y[labeled_idx])
    
    probs = model.predict_proba(X[unlabeled_idx])
    uncertainty = np.abs(probs[:, 0] - 0.5)
    
    query_idx = np.argmin(uncertainty)
    selected = unlabeled_idx[query_idx]
    
    labeled_idx.append(selected)
    unlabeled_idx.remove(selected)

print("Final labeled indices:", labeled_idx)
```

```python
# 5 (a)
from itertools import combinations

transactions = [
    {'A', 'B'},
    {'A', 'C'},
    {'B', 'C'},
    {'A', 'B', 'C'}
]

items = sorted(set().union(*transactions))

C1 = [{item} for item in items]
C2 = [set(c) for c in combinations(items, 2)]
C3 = [set(c) for c in combinations(items, 3)]


# 5 (b)
def get_support(itemset):
    return sum(1 for t in transactions if itemset.issubset(t))

support_C1 = {tuple(i): get_support(set(i)) for i in C1}
support_C2 = {tuple(i): get_support(set(i)) for i in C2}
support_C3 = {tuple(i): get_support(set(i)) for i in C3}


# 5 (c)
min_support = 2

L1 = {k: v for k, v in support_C1.items() if v >= min_support}
L2 = {k: v for k, v in support_C2.items() if v >= min_support}
L3 = {k: v for k, v in support_C3.items() if v >= min_support}

print("Frequent 1-itemsets:", L1)
print("Frequent 2-itemsets:", L2)
print("Frequent 3-itemsets:", L3)
```

```python
# 6 (a)
transactions = [
    {'A', 'B'},
    {'A', 'C'},
    {'B', 'C'}
]

tid_list = {}
for tid, transaction in enumerate(transactions):
    for item in transaction:
        if item not in tid_list:
            tid_list[item] = set()
        tid_list[item].add(tid)

print("TID List:", tid_list)


# 6 (b)
from itertools import combinations

intersections = {}
items = list(tid_list.keys())

for r in range(2, len(items) + 1):
    for combo in combinations(items, r):
        intersect = set.intersection(*(tid_list[i] for i in combo))
        intersections[combo] = intersect

print("\nIntersections:", intersections)


# 6 (c)
min_support = 2

frequent_itemsets = {}

for item, tids in tid_list.items():
    if len(tids) >= min_support:
        frequent_itemsets[(item,)] = tids

for combo, tids in intersections.items():
    if len(tids) >= min_support:
        frequent_itemsets[combo] = tids

print("\nFrequent Itemsets:", frequent_itemsets)
```

```python
# 6 (a)
import numpy as np
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from scipy.stats import mode

X, y = make_classification(n_samples=400, n_features=4, random_state=42)

n_models = 10
models = []
bootstrap_samples = []

for _ in range(n_models):
    indices = np.random.choice(len(X), size=len(X), replace=True)
    X_sample = X[indices]
    y_sample = y[indices]
    bootstrap_samples.append((X_sample, y_sample))


# 6 (b)
for X_sample, y_sample in bootstrap_samples:
    model = DecisionTreeClassifier()
    model.fit(X_sample, y_sample)
    models.append(model)


# 6 (c)
predictions = []

for model in models:
    preds = model.predict(X)
    predictions.append(preds)

predictions = np.array(predictions)
final_preds = mode(predictions, axis=0).mode.flatten()


# 6 (d)
accuracy = accuracy_score(y, final_preds)
print("Bagging Accuracy:", accuracy)
```

```python
# 12 (a)
import numpy as np
import random

epsilon = 0.2
actions = [0, 1]  # example actions
q_values = np.zeros(len(actions))


# 12 (b)
explore_count = 0
exploit_count = 0
iterations = 20

for _ in range(iterations):
    if random.random() < epsilon:
        action = random.choice(actions)
        explore_count += 1
    else:
        action = np.argmax(q_values)
        exploit_count += 1
    
    reward = random.random()
    q_values[action] += 0.1 * (reward - q_values[action])


# 12 (c)
exploration_steps = explore_count
exploitation_steps = exploit_count


# 12 (d)
print("Exploration steps:", exploration_steps)
print("Exploitation steps:", exploitation_steps)
print("Final Q-values:", q_values)
```

```python
# 16 (a)
import numpy as np

states = [0, 1, 2]
actions = ['Left', 'Right']

V = np.zeros(len(states))
gamma = 0.9


# 16 (b)
def get_next_state_reward(s, a):
    if a == 'Left':
        next_state = max(0, s - 1)
    else:
        next_state = min(2, s + 1)
    reward = 1 if next_state == 2 else 0
    return next_state, reward

theta = 1e-4

while True:
    delta = 0
    new_V = V.copy()
    
    for s in states:
        v = 0
        for a in actions:
            next_state, reward = get_next_state_reward(s, a)
            v += 0.5 * (reward + gamma * V[next_state])
        
        new_V[s] = v
        delta = max(delta, abs(V[s] - new_V[s]))
    
    V = new_V
    print("Updated V:", V)
    
    if delta < theta:
        break


# 16 (c)
print("\nFinal Value Function:", V)
```

```python
# 19 (a)
import random
import numpy as np

states = [0, 1, 2]
actions = ['Left', 'Right']
gamma = 0.9

def get_next_state_reward(s, a):
    if a == 'Left':
        next_state = max(0, s - 1)
    else:
        next_state = min(2, s + 1)
    reward = 1 if next_state == 2 else 0
    return next_state, reward

episodes = []

for _ in range(5):
    episode = []
    state = random.choice(states)
    
    for _ in range(5):
        action = random.choice(actions)
        next_state, reward = get_next_state_reward(state, action)
        episode.append((state, reward))
        state = next_state
    
    episodes.append(episode)


# 19 (b)
returns = []

for episode in episodes:
    G = 0
    episode_returns = []
    
    for t in reversed(range(len(episode))):
        _, reward = episode[t]
        G = reward + gamma * G
        episode_returns.insert(0, G)
    
    returns.append(episode_returns)


# 19 (c)
value_sum = {s: 0 for s in states}
value_count = {s: 0 for s in states}

for episode, Gs in zip(episodes, returns):
    for (state, _), G in zip(episode, Gs):
        value_sum[state] += G
        value_count[state] += 1

V = {s: value_sum[s] / value_count[s] if value_count[s] > 0 else 0 for s in states}

print("Episodes:", episodes)
print("\nReturns:", returns)
print("\nEstimated Value Function:", V)
```

```python
# 20 (a)
import numpy as np
import random

states = [0, 1, 2]
actions = ['Left', 'Right']

Q = np.zeros((len(states), len(actions)))

alpha = 0.1
gamma = 0.9


# 20 (b) (c) (d)
def get_next_state_reward(s, a):
    if a == 'Left':
        next_state = max(0, s - 1)
    else:
        next_state = min(2, s + 1)
    reward = 1 if next_state == 2 else 0
    return next_state, reward

episodes = 20

for _ in range(episodes):
    state = random.choice(states)
    
    for _ in range(10):
        action_idx = random.choice([0, 1])
        action = actions[action_idx]
        
        next_state, reward = get_next_state_reward(state, action)
        
        best_next = np.max(Q[next_state])
        
        Q[state, action_idx] += alpha * (reward + gamma * best_next - Q[state, action_idx])
        
        state = next_state


# 20 (e)
print("Final Q-table:\n", Q)
```
